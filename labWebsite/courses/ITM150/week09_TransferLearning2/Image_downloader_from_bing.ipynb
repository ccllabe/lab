```python
import requests
from bs4 import BeautifulSoup
import os

def download_images(keyword, num_images):

    if not os.path.exists(keyword):
        os.makedirs(keyword)

    url = f"https://www.bing.com/images/search?q={keyword}&form=HDRSC2"

    response = requests.get(url)
    soup = BeautifulSoup(response.text, 'html.parser')

    image_links = soup.find_all('img', class_='mimg')

    for i, image_link in enumerate(image_links[:num_images]):
        try:
            image_url = image_link['src']
            image_content = requests.get(image_url).content
            with open(os.path.join(keyword, f"{keyword}_{i+1}.jpg"), 'wb') as f:
                f.write(image_content)
            print(f"Image {i+1} downloaded successfully.")
        except Exception as e:
            print(f"Error downloading image {i+1}: {str(e)}")

if __name__ == "__main__":
    keyword = input("Enter keyword for image search: ")
    num_images = int(input("Enter number of images to download: "))
    download_images(keyword, num_images)
```
